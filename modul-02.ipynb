### Membuka Rekap Data: Membaca File CSV dengan Pandas

Sekarang kasus intinya. Data PODES kelurahan Bandar Lampung tersimpan sebagai berkas **CSV** (*comma-separated values*): berkas teks biasa yang tiap barisnya adalah satu catatan dan tiap kolomnya dipisahkan koma — bentuk paling sederhana dari **data terstruktur**, data yang tertata dalam baris dan kolom bermakna tetap. Baris pertamanya berisi nama-nama kolom. Kalau dibuka dengan Notepad, isinya kira-kira:

```
kode_prov,nama_prov,kode_kab,nama_kab,kode_kec,nama_kec,kode_des,nama_des,...
18,Lampung,71,Kota Bandar Lampung,10,Teluk Betung Barat,5,Bakung,...
```

Membaca teks seperti itu dengan mata tentu melelahkan. Di sinilah **Pandas** bekerja: ia memuat CSV menjadi tabel rekapitulasi yang rapi — persis lembar rekap kompilasi data studio di Excel — yang di Pandas disebut ***DataFrame***.

Berkas kita berisi 34 kolom, padahal pertemuan ini hanya butuh sebagian. Maka sebelum membaca, kita siapkan dua variabel penampung: `path_data` untuk alamat berkasnya (ingat, notebook kita berada di folder `modul`, sehingga *path* relatifnya naik satu tingkat dulu dengan `..`) dan `kolom_pilihan` untuk daftar kolom yang ingin dibaca. Dengan begitu, perintah bacanya sendiri tetap pendek:

````python
import pandas as pd

path_data = "../dataset-ganjil-2026-2027/bdl_kelurahan.csv"
kolom_pilihan = ["nama_kec", "nama_des", "status_daerah",
                 "kepadatan_pddk", "jml_tk_negeri", "jml_tk_swasta",
                 "jml_ra_ba_negeri", "jml_ra_ba_swasta"]

df = pd.read_csv(path_data, usecols=kolom_pilihan)
df.head()
````

Tiga baris pekerjaan terjadi di sana. `import pandas as pd` memanggil *library* Pandas dari rak **environment** kita dan memberinya nama panggilan pendek `pd` — cukup sekali di awal notebook. `pd.read_csv(...)` membaca berkas pada `path_data`, mengambil hanya kolom-kolom dalam `kolom_pilihan` (lewat `usecols=`), lalu hasilnya kita tampung di variabel `df`. Terakhir, `df.head()` mengintip **lima baris pertama** tabel untuk memastikan bacaan kita benar.

::: {.callout-note title="Konsep Python"}
### Tipe Data *list* {.unnumbered .unlisted}
`kolom_pilihan` adalah sebuah ***list***: tipe data yang menampung **banyak nilai sekaligus dalam satu kotak, berurutan**. Penulisannya memakai kurung siku, dengan koma sebagai pemisah: `["nama_kec", "nama_des", ...]`. Kalau variabel biasa adalah satu kotak berlabel, **list** adalah **rak berlabel yang isinya bernomor urut**.

Nomor urut itu disebut **indeks**, dan — ini kekhasan Python — **dihitung mulai dari 0**: `kolom_pilihan[0]` berisi `"nama_kec"`, `kolom_pilihan[1]` berisi `"nama_des"`, dan seterusnya. Panjang **list** bisa ditanyakan dengan `len(kolom_pilihan)`.

**list** akan sangat sering kita pakai untuk menampung daftar nama kolom seperti ini — daripada mengetik daftar panjang berulang-ulang di banyak perintah, kita tulis sekali di satu variabel lalu sebut labelnya.
:::

### Kelurahan vs Kecamatan: Membedakan Unit Amatan dan Unit Analisis

Perhatikan lagi: setiap baris tabel kita adalah satu **kelurahan** — pada level itulah BPS mencatat datanya. Dalam bahasa metodologi, kelurahan adalah **unit amatan** (*unit of observation*) dataset ini: satuan tempat data dicatat, seperti satu lembar kuesioner terisi untuk satu kelurahan.

Namun pertanyaan perencanaan tidak selalu dijawab pada level pencatatan. Ketika pimpinan bertanya "kecamatan mana yang paling minim fasilitas PAUD?", kesimpulannya diambil pada level **kecamatan** — itulah **unit analisis** (*unit of analysis*): satuan tempat kesimpulan dibuat, seperti lembar rekapitulasi per wilayah yang merangkum tumpukan kuesioner. Unit amatan dan unit analisis boleh sama, boleh berbeda; yang fatal adalah tidak sadar sedang bekerja di level yang mana.

Mari kenali kedua level itu di data kita. Satu kolom di *DataFrame* dipanggil dengan `df["nama_kolom"]`, dan `nunique()` menghitung banyaknya **nilai unik** (nilai yang berbeda-beda) di dalamnya:

````python
print(f"Jumlah kelurahan (unit amatan) : {df['nama_des'].nunique()}")
print(f"Jumlah kecamatan               : {df['nama_kec'].nunique()}")
````

126 kelurahan tersebar di 20 kecamatan. Lalu berapa kelurahan yang dinaungi masing-masing kecamatan? `value_counts()` menghitung **frekuensi kemunculan** tiap nilai pada sebuah kolom — alat andalan untuk meringkas **data kategorikal** (data berupa golongan/kategori, bukan angka hitungan):

````python
df["nama_kec"].value_counts()
````

Tabel kecil itu sudah merupakan rekapitulasi pada unit analisis kecamatan: Teluk Betung Selatan menaungi paling banyak kelurahan, dan seterusnya. Simpan baik-baik `value_counts()` di ingatan — di Latihan Mandiri ia akan dipakai untuk meringkas kategori yang lebih bermakna sosial-ekonomi, dan di Modul 3 kita akan membuat rekapitulasi per kecamatan yang lebih kaya menggunakan SQL.

### Merapikan Tabel: Membuat, Menyusun Ulang, dan Menghapus Kolom

Fasilitas pendidikan usia dini di data kita terpecah menjadi empat kolom: TK negeri, TK swasta, RA/BA negeri, dan RA/BA swasta (RA/BA adalah *raudatul athfal*/*bustanul athfal*, PAUD di bawah Kemenag). Untuk pertanyaan alokasi bantuan, yang lebih berguna adalah **totalnya**. Di lembar rekap studio, kita akan menambahkan satu kolom jumlahan di ujung tabel — di Pandas pun sama: menugaskan nilai ke nama kolom yang belum ada akan **membuat kolom baru**:

````python
df["jml_paud"] = df["jml_tk_negeri"] + df["jml_tk_swasta"] + df["jml_ra_ba_negeri"] + df["jml_ra_ba_swasta"]
df.head(3)
````

Perhatikan dua hal: penjumlahan antar-kolom bekerja **serentak untuk seluruh 126 baris** tanpa kita menghitung satu-satu, dan kolom baru `jml_paud` muncul di **ujung kanan** tabel. Posisi di ujung kanan itu kurang enak dibaca — kolom sepenting itu layak berdiri di dekat identitas kelurahan. Menyusun ulang posisi kolom caranya sederhana: sodorkan ***list*** berisi urutan kolom yang kita mau (pemakaian ulang konsep *list* barusan):

````python
urutan_kolom = ["nama_kec", "nama_des", "status_daerah",
                "kepadatan_pddk", "jml_paud",
                "jml_tk_negeri", "jml_tk_swasta",
                "jml_ra_ba_negeri", "jml_ra_ba_swasta"]

df = df[urutan_kolom]
df.head(3)
````

Terakhir, setelah totalnya ada, empat kolom pecahannya tidak kita perlukan lagi hari ini. Membiarkan kolom tak terpakai membuat tabel lebar dan mata cepat lelah — seperti membiarkan kertas kalkir bekas menumpuk di meja gambar. Kita buang dengan `drop()`, lagi-lagi lewat variabel *list* agar rapi:

````python
kolom_dibuang = ["jml_tk_negeri", "jml_tk_swasta", "jml_ra_ba_negeri", "jml_ra_ba_swasta"]

df = df.drop(columns=kolom_dibuang)
df.head(3)
````

### Menyaring Baris: Filtrasi Data Kelurahan Berdasarkan Kriteria Tertentu

Tabel sudah rapi; kini saatnya mengajukan pertanyaan sang pemberi kasus: **kelurahan mana yang paling membutuhkan bantuan fasilitas PAUD?** Tafsiran pertama yang paling sederhana: kelurahan yang fasilitas PAUD-nya sangat sedikit — katakanlah satu atau tidak ada sama sekali.

Menyaring baris di Pandas polanya `df[kondisi]`: "dari tabel `df`, ambil baris-baris yang memenuhi kondisi". Kondisinya kita tulis sebagai perbandingan atas sebuah kolom:

````python
kelurahan_minim_paud = df[df["jml_paud"] <= 1]
print(f"Jumlah kelurahan dengan PAUD <= 1: {len(kelurahan_minim_paud)}")
kelurahan_minim_paud.head()
````

::: {.callout-note title="Konsep Python"}
### Operator Perbandingan {.unnumbered .unlisted}
`df["jml_paud"] <= 1` memakai **operator perbandingan** — tanda yang membandingkan dua nilai dan menjawab benar (`True`) atau salah (`False`):

| Operator | Arti | Contoh |
|---|---|---|
| `==` | sama dengan | `status == 1` |
| `!=` | tidak sama dengan | `nama != "Bakung"` |
| `>` | lebih besar | `kepadatan > 10000` |
| `<` | lebih kecil | `luas < 5` |
| `>` | lebih besar atau sama | `jml_paud >= 2` |
| `<=` | lebih kecil atau sama | `jml_paud <= 1` |

Perhatikan: **sama dengan ditulis dua kali** (`==`), karena tanda `=` tunggal sudah dipakai untuk penugasan variabel (Modul 1). Menukar keduanya adalah salah ketik yang sangat umum.

Saat dipakai pada kolom **DataFrame**, perbandingan diuji **serentak ke seluruh baris**, menghasilkan deretan `True`/`False` sepanjang tabel — dan `df[...]` menahan hanya baris-baris yang `True`. Itulah mekanika di balik penyaringan barusan.
:::


````python
kriteria_padat = df["kepadatan_pddk"] > 10000
kriteria_minim = df["jml_paud"] <= 1

kelurahan_prioritas = df[kriteria_padat & kriteria_minim]
print(f"Jumlah kelurahan prioritas: {len(kelurahan_prioritas)}")
kelurahan_prioritas
````

::: {.callout-note title="Konsep Python"}
### Operator Boolean `&` dan `|` {.unnumbered .unlisted}
Untuk menggabungkan dua kondisi, Python menyediakan **operator boolean** (dinamai dari aljabar *Boolean*, aljabar benar-salah):

- `&` berarti **dan** — hasilnya `True` hanya jika **kedua** kondisi benar.

- `|` berarti **atau** — hasilnya `True` jika **salah satu saja** kondisi benar.

Dalam Python murni kata kuncinya `and` dan `or`, tetapi **khusus untuk kolom Pandas kita wajib memakai lambang `&` dan `|`**, karena pengujian dilakukan serentak ke ratusan baris sekaligus. Satu aturan lagi: jika kondisinya ditulis langsung tanpa variabel penampung, tiap kondisi **harus dikurung**, misalnya `df[(df["kepadatan_pddk"] > 10000) & (df["jml_paud"] <= 1)]` — tanpa kurung, Python salah membaca urutannya. Menampung kondisi ke variabel seperti yang kita lakukan membuat kurung tak diperlukan dan kodenya jauh lebih terbaca.
:::
