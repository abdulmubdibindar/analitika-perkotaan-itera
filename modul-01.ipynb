### Padanan Path dan Struktur Folder Tree {.unnumbered .unlisted}

Struktur folder tree di atas dapat direpresentasikan ke dalam bentuk **path**. Sebagai contoh, struktur folder tree berikut:

```
C:
└── Users
    └── Aulia
        └── Documents
            └── analitika-perkotaan
```

akan dituliskan sebagai path absolut: `C:/Users/Aulia/Documents/analitika-perkotaan`.



**Mengapa pemahaman path ini penting?**

Perhatikan contoh penamaan direktori dengan path yang salah di bawah ini:

```
C:
└── Users
    └── Aulia
        ├── Documents
        │   └── analitika-perkotaan
        └── Music
            └── analitika-perkotaan
```

Meskipun kedua folder sama-sama bernama `analitika-perkotaan`, path absolutnya sangat berbeda: yang satu berada di `C:/Users/Aulia/Documents/analitika-perkotaan` dan yang satu lagi di `C:/Users/Aulia/Music/analitika-perkotaan`. Jika program kita salah mencari lokasi, ia tidak akan menemukan berkas yang tepat.

:::



::: {.callout-note title="Pengetahuan Komputer"}

### Penggunaan Simbol Slash di Windows vs Unix untuk penulisan *Path* {.unnumbered .unlisted}

Sistem operasi Windows secara tradisional menggunakan simbol *backslash* (`\`) untuk memisahkan nama folder di dalam path (misalnya `C:\Users\Aulia`). Namun, sistem berbasis Unix (seperti Linux dan macOS), serta browser web, menggunakan *forward slash* (`/`) (misalnya `/Users/Aulia`). Dalam pemrograman Python, disarankan untuk menggunakan *forward slash* (`/`) karena Python dapat menerjemahkannya secara otomatis di Windows dan mencegah galat akibat karakter khusus (*escape character*).

:::



Susunan bercabang seperti bagan di atas disebut ***folder tree*** (pohon direktori). Untuk menunjuk lokasi sebuah berkas di dalam pohon itu, komputer memakai ***path***: rangkaian nama folder yang dilalui untuk sampai ke berkas tersebut, dipisahkan garis miring. Ada dua cara menuliskannya:



- ***Path* lengkap (absolut)** — dari pangkal pohon sampai ujung, misalnya `C:/Users/Aulia/Documents/analitika-perkotaan/skrip/sapa.py`. Seperti alamat surat lengkap: kota, kecamatan, jalan, nomor rumah.
- ***Path* relatif** — dihitung dari posisi kita berdiri sekarang. Jika kita sedang berada di dalam `analitika-perkotaan`, berkas yang sama cukup ditunjuk dengan `skrip/sapa.py`. Seperti berkata "dua rumah di sebelah kanan" kepada orang yang sudah berdiri di jalan yang sama.



Sepanjang semester kita akan lebih sering memakai *path* relatif, dan karena itu **posisi kita berdiri** (folder mana yang sedang terbuka) menjadi penting. Ini akan segera terasa saat kita bekerja dengan terminal.




### Menyiapkan Ruang Kerja: *Environment* Mandiri dan Instalasi Conda

Di studio, setiap perencana punya ruang kerja studio pribadi berisi meja, kursi, dan rak peralatan miliknya sendiri, supaya pengaturan alat untuk proyek yang satu tidak tercampur dengan proyek lain. Dalam dunia Python, ruang kerja pribadi itu bernama ***environment***, dan petugas yang menyiapkannya bernama **Conda**. Satu komputer bisa memiliki banyak *environment* sekaligus, masing-masing berisi versi Python dan *library* yang berbeda, tanpa saling mengganggu.

**Langkah nol: periksa dulu penghuni lama ruangan.** Sebelum menata meja kerja, peserta studio memeriksa dulu meja siapa saja yang sudah menumpuk di ruangan itu — supaya tidak keliru memakai meja orang lain, dan tidak menggeser barang yang ternyata masih dipakai. Di komputer, langkah itu berarti mendata Python apa saja yang **sudah ada** sebelum kita menambah apa pun. Buka **Command Prompt** atau **PowerShell**, lalu ketik dua perintah berikut satu per satu. Tuliskan persis seperti tertulis, lengkap dengan akhiran `.exe`-nya:

```bash
where.exe python
where.exe conda
```

Perintah `where.exe` menampilkan **semua** lokasi program bernama itu yang ditemukan komputer, berurutan dari yang paling didahulukan. Salin hasilnya ke buku catatan Anda — kita akan menagihnya lagi di langkah kelima. Jangan kaget bila keluarannya lebih dari satu baris; justru itu yang biasa terjadi.

::: {.callout-warning title="Penting!"}
### Mengapa Ditulis `where.exe`, Bukan `where` {.unnumbered .unlisted}
Windows menyediakan dua terminal yang tampilannya nyaris serupa: **Command Prompt** dan **PowerShell**. Bedakan dari tulisan di awal barisnya — PowerShell selalu mengawali barisnya dengan `PS`, seperti `PS C:\Users\Nama>`, sedangkan Command Prompt langsung `C:\Users\Nama>`. Perhatikan ini baik-baik, karena pada Windows 11 PowerShell-lah yang paling sering terbuka lebih dulu — termasuk saat kita membuka terminal bawaan VS Code nanti.

Perbedaannya berakibat langsung di langkah ini. Di Command Prompt, `where` adalah nama program pencari lokasi yang kita maksud. Di PowerShell, nama `where` sudah dipakai lebih dulu sebagai singkatan perintah bawaannya sendiri (`Where-Object`) yang tugasnya menyaring daftar — sama sekali tidak berhubungan dengan mencari lokasi program.

Bentuk kegagalannya justru yang paling menyesatkan: PowerShell **tidak menampilkan pesan galat apa pun**. Ia hanya diam, kursor kembali ke baris kosong, seolah-olah perintahnya sudah dikerjakan. Layar kosong itu sangat mudah dibaca sebagai baris terakhir tabel di bawah — "belum ada Python" — padahal komputer Anda bisa jadi punya tiga sekaligus. Berbeda dengan pesan `'...' is not recognized` yang jelas-jelas menyuruh kita memperbaiki sesuatu, kegagalan yang diam justru membuat kita melangkah ke tahap berikutnya dengan kesimpulan yang keliru.

Akhiran `.exe` menutup celah itu: ia memaksa terminal menjalankan program pencarinya, bukan singkatan tadi. Bentuk `where.exe` **aman dipakai di ketiganya** — Command Prompt, PowerShell, maupun Anaconda Prompt — jadi biasakan menulis akhiran itu sepanjang mata kuliah ini.
:::

Sekarang cocokkan keluaran Anda dengan tabel berikut, supaya langkah nol ini menghasilkan sebuah **kesimpulan**, bukan sekadar tumpukan tulisan:

| Yang muncul di keluaran | Artinya |
| :--- | :--- |
| `C:\Python27\ArcGIS10.8\...` atau `...\ArcGISx6410.8\...` | ArcGIS Desktop 10.x — membawa Python 2.7 |
| `...\ArcGIS\Pro\bin\Python\...` (pada `where.exe conda`) | ArcGIS Pro — membawa Conda sendiri |
| `...\WindowsApps\python.exe` | Python dari Microsoft Store |
| `C:\Python3xx\...` atau `...\AppData\Local\Programs\Python\...` | Python dari situs python.org |
| tidak ada keluaran sama sekali | belum ada Python — kondisi paling bersih; pastikan dulu yang Anda ketik `where.exe`, bukan `where` |

::: {.callout-warning}
### Komputer Anda Mungkin Sudah Punya Python Tanpa Anda Sadari {.unnumbered .unlisted}
Bila di komputer Anda terpasang **ArcGIS**, maka Python **sudah ada di sana sejak lama** — dan ada dua kemungkinan yang penanganannya berbeda. Tabel di atas menunjukkan Anda termasuk yang mana. Baca **hanya kasus Anda**; bila keduanya muncul, kerjakan keduanya.

**Kasus A — ArcGIS Desktop 10.x (ArcMap).** Yang dibawanya adalah Python **versi 2.7**, umumnya di `C:/Python27/ArcGIS10.8`, dan ia sama sekali tidak memakai Conda. Apakah Python 2.7 ini terdaftar lebih dulu di daftar alamat komputer **tidak bisa ditebak dari luar**: pada sebagian mesin ia menang, pada sebagian lain ia tidak terdaftar sama sekali meski ArcGIS-nya terpasang. Bila ia menang, mengetik `python` di terminal baru menjalankan **Python 2.7** — bukan Python kita. Bahayanya justru bukan karena ia menolak bekerja, melainkan sebaliknya: ia mau berjalan, lalu gagal di tempat yang membingungkan. Kode seperti `print(f"Halo, {nama}")` yang akan kita tulis sebentar lagi ditolaknya dengan `SyntaxError`, padahal kode itu benar. Anda akan mengira diri sendiri salah ketik, padahal yang keliru adalah Python yang menjalankannya. Inilah alasan langkah nol dikerjakan sebelum apa pun.

**Kasus B — ArcGIS Pro.** ArcGIS Pro membawa **Conda-nya sendiri**, lengkap dengan sebuah *environment* bernama `arcgispro-py3` di `C:/Program Files/ArcGIS/Pro/bin/Python/envs/arcgispro-py3`. Bahayanya lebih halus dan baru terasa di langkah pertama nanti: perintah `conda` yang dikenali komputer Anda bisa jadi **milik ArcGIS Pro**, sehingga ruang kerja mata kuliah ini tanpa sadar dibangun di dalam instalasi ArcGIS. Periksa keluaran `where.exe conda` Anda: bila ada baris yang memuat potongan `\ArcGIS\Pro\`, jangan pakai terminal "Python Command Prompt" bawaan ArcGIS — gunakan **Anaconda Prompt** milik Miniconda. Bila Anaconda Prompt belum ada di komputer Anda, pasang Miniconda dari berkas yang dibagikan dosen terlebih dahulu.

**Berlaku untuk kedua kasus: jangan dihapus, dan jangan dipakai untuk mata kuliah ini.** Jangan dihapus, karena ArcGIS bergantung sepenuhnya pada Python bawaannya — merusaknya berarti merusak ArcGIS Anda. Jangan dipakai, dan ini bukan soal selera: Python 2.7 pada Kasus A sudah berhenti dikembangkan sejak 1 Januari 2020 sehingga tidak satu pun *library* yang kita pakai semester ini berjalan di sana, sedangkan *environment* bawaan ArcGIS Pro pada Kasus B memang dirancang untuk tidak diubah-ubah. *Environment* yang kita buat hari ini berdiri **berdampingan** dengan Python-Python itu: menambah satu ruang kerja baru, tanpa menggusur ruang kerja yang sudah ada.

Di luar dua kasus itu, Python juga bisa masuk lewat **Microsoft Store** atau pemasangan dari situs **python.org**, sering kali tanpa kita sadari. Jangan buru-buru menganggapnya sepele: pada komputer yang sudah diperiksa untuk mata kuliah ini, justru Python dari python.org inilah yang **memenangkan baris pertama** daftar alamat — mengalahkan Python bawaan ArcGIS sekalipun. Bahayanya berkebalikan dengan Kasus A, dan justru karena itu lebih licin: ia **tidak menolak apa pun**. Kode Anda jalan, tidak ada tulisan merah, tetapi *library* yang dipakainya berbeda versi sehingga angka yang keluar bisa berbeda dari yang muncul di layar teman sebelah — tanpa satu pun pesan yang memberi tahu. Kegagalan yang diam jauh lebih sulit dilacak daripada yang berisik. Karena itu keberadaan mereka tidak cukup diingat, melainkan harus dibuktikan siapa yang sesungguhnya berjalan — dan itulah yang kita kerjakan di langkah kelima nanti.
:::

**Langkah pertama: periksa dulu, jangan langsung memasang.** Komputer laboratorium kampus sudah dipasangi Conda (bahkan lengkap dengan *environment* mata kuliah ini) oleh tim dosen, jadi memasang ulang hanya membuang waktu. Buka **Anaconda Prompt**.

::: {.callout-warning title="Penting!"}
### Konsep Path & Menjalankan Conda di Terminal Biasa {.unnumbered .unlisted}
Anaconda Prompt adalah terminal khusus yang sudah dikonfigurasi secara otomatis agar dapat mengenali perintah `conda`. Jika Anda ingin menjalankan perintah `conda` dari Command Prompt biasa atau PowerShell, Anda harus menginisialisasinya terlebih dahulu dengan menjalankan perintah `conda init` di Anaconda Prompt, lalu menutup dan membuka kembali terminal biasa Anda.

Setelah memahami hal ini, cobalah buka Command Prompt biasa atau PowerShell di komputer Anda, lalu ketik `conda --version` untuk memverifikasi apakah ia sudah mengenali perintah tersebut.
:::

Di Anaconda Prompt, ketik:

```bash
conda --version
```

Ada dua kemungkinan jawaban:

- Muncul tulisan seperti `conda 24.11.0` (angkanya boleh berbeda) — Conda **sudah terpasang**. Lanjutkan ke pemeriksaan *environment* di bawah.

- Muncul pesan seperti `'conda' is not recognized ...` — Conda **belum ada**. Pasang **Miniconda** dari berkas instalasi yang dibagikan dosen (tersedia *offline*, tidak perlu mengunduh), ikuti pilihan bawaannya sampai selesai, tutup terminalnya, buka **Anaconda Prompt** yang baru, lalu ulangi perintah di atas sampai versi Conda muncul.

**Langkah kedua: periksa apakah *environment* mata kuliah sudah ada.** Ketik:

```bash
conda env list
```

Perintah ini menampilkan daftar semua ruang kerja (*environment*) yang ada di komputer. Jika di daftarnya sudah ada baris bernama `analitika-perkotaan` (kondisi normal di komputer lab), ruang kerja kita sudah berdiri — lompat ke langkah keempat.

**Langkah ketiga (hanya jika belum ada): buat *environment* dari berkas resep.** Dosen membagikan sebuah berkas bernama `environment.yml`. Salin berkas itu ke direktori proyek `analitika-perkotaan` yang kita buat tadi. Isinya seperti ini:

```yaml
name: analitika-perkotaan

channels:
  - conda-forge
  - nodefaults

dependencies:
  # Dasar: Modul 2-4
  - python=3.12
  - pandas
  - openpyxl

  # Visual: Modul 5
  - matplotlib-base
  - seaborn

  # Spasial: Modul 6-7
  - geopandas
  - leafmap
  - mapclassify
  - folium
  - contextily

  # Penarikan data: Modul 9
  - requests
  - beautifulsoup4
  - osmnx

  # Pemodelan: Modul 10
  - scikit-learn

  # Dashboard: Tugas Besar Pekan 11-16 & UAS
  - streamlit
  - streamlit-folium

  # Infrastruktur agar notebook berjalan
  - ipykernel
  - jupyterlab
  - notebook
  - nbformat
  - nbclient
  - pyyaml

  # Bantu: pembacaan dataset dan tabel interaktif
  - dbfread
  - itables

variables:
  PYTHONNOUSERSITE: "1"
```

Berkas ini adalah **daftar belanja perabot** untuk membangun ruang kerja satu semester: nama ruang kerjanya (`name`), toko tempat membelinya (`channels`), daftar rak, kursi, serta meja yang dibeli (`dependencies`), dan satu aturan rumah yang berlaku di dalam ruangan itu (`variables`). Baris berawalan tanda `#` adalah **komentar**: keterangan untuk manusia, diabaikan komputer. Aturan rumah `PYTHONNOUSERSITE` itu melarang ruang kerja kita mengambil perabot dari sebuah rak titipan bersama di luar ruangan — alasannya dibahas pada *callout* terakhir bagian ini. Dengan satu berkas ini kita tidak perlu mengetik nama *library* satu per satu — cukup satu perintah, dan Conda membelanjakan semuanya:

```bash
cd Documents\analitika-perkotaan
conda env create -f environment.yml
```

Perintah `cd` (*change directory*) memindahkan posisi berdiri kita ke direktori proyek — ingat pembahasan *path* tadi: `environment.yml` ditunjuk dengan *path* relatif, sehingga kita harus berdiri di folder tempat berkas itu berada. Proses ini butuh beberapa menit.

**Langkah keempat: masuk ke ruang kerja.** Aktifkan *environment* dengan:

```bash
conda activate analitika-perkotaan
```

Perhatikan awalan tulisan di terminal berubah dari `(base)` menjadi `(analitika-perkotaan)`. Awalan itu adalah papan nama ruangan: ia memberi tahu ruang kerja mana yang sedang kita buka pintunya untuk mulai bekerja. Mulai sekarang, setiap kali membuka terminal untuk mata kuliah ini, biasakan memeriksa papan nama tersebut.

**Langkah kelima: buktikan ruang kerjanya benar-benar terbuka.** Papan nama `(analitika-perkotaan)` itu **klaim**, bukan bukti. Papan nama bisa menyala sementara yang benar-benar bekerja adalah Python lain — dan ini bukan kemungkinan teoretis, melainkan kejadian yang wajar di komputer yang sudah punya banyak Python (ingat hasil langkah nol Anda). Karena itu, sekali di awal semester, kita buktikan. Dengan papan nama `(analitika-perkotaan)` menyala, jalankan tiga perintah berikut — perhatikan lagi akhiran `.exe` pada perintah pertama:

```bash
where.exe python
python -c "import sys; print(sys.prefix)"
python -c "import sys; print(sys.path)"
```

Ruang kerja Anda dinyatakan **lulus** bila ketiga syarat ini terpenuhi:

1. Baris **paling atas** keluaran `where.exe python` memuat potongan `envs\analitika-perkotaan`.
2. Keluaran perintah kedua berakhir dengan `envs\analitika-perkotaan`.
3. Keluaran perintah ketiga **tidak memuat** potongan `AppData\Roaming\Python` di mana pun.

Catat ketiga keluaran itu — Latihan Mandiri butir 5 akan menagihnya.

::: {.callout-warning}
### Jika Ruang Kerja Anda Tidak Lulus {.unnumbered .unlisted}
**Perintah pertama tidak mengeluarkan apa pun**: hampir pasti yang terketik `where`, bukan `where.exe`, dan terminal Anda PowerShell. Ulangi dengan akhiran `.exe`-nya.

**Syarat 1 atau 2 gagal** (yang muncul Python lain): tutup terminal, buka **Anaconda Prompt** baru, jalankan `conda activate analitika-perkotaan`, lalu ulangi. Bila masih gagal, kemungkinan besar *environment*-nya belum benar-benar terbentuk — ulangi langkah kedua dan ketiga. Bila yang muncul di baris teratas justru `C:\Python27\...`, itu Kasus A pada *callout* sebelumnya — penyebabnya hampir selalu `conda activate` yang belum dijalankan di terminal tersebut.

**Syarat 3 gagal** (muncul `AppData\Roaming\Python`): **jangan diperbaiki sendiri, laporkan ke dosen.** Kondisi ini berarti ada rak paket titipan yang membayangi ruang kerja Anda, dan membersihkannya sembarangan bisa merusak program lain di komputer Anda — termasuk ArcGIS Pro bila Anda memakainya.
:::

::: {.callout-note title="Pengetahuan Komputer"}
### Satu Komputer, Banyak Python: Mengapa Harus Dibuktikan {.unnumbered .unlisted}
Ketika kita mengetik `python` di terminal, komputer **tidak tahu** Python mana yang kita maksud. Ia hanya menyusuri sebuah daftar alamat bernama **PATH** dari atas ke bawah, lalu menjalankan **program pertama** bernama `python` yang ia temukan. Itulah sebabnya `where.exe python` bisa menampilkan beberapa baris sekaligus: baris pertama adalah yang menang, sisanya menganggur.

`conda activate` bekerja dengan cara menyelipkan alamat *environment* kita ke **urutan paling atas** daftar itu. Bila ada sesuatu yang menyelipkan alamat lain lebih atas lagi, papan nama tetap menyala tetapi yang berjalan Python yang lain. Analogi studionya: papan nama di pintu ruangan bisa saja tertulis nama Anda, sementara yang duduk mengerjakan gambar di dalamnya orang lain. Satu-satunya cara memastikan adalah menengok ke dalam — dan `where.exe python` beserta `sys.prefix` adalah cara menengoknya.

Empat baris barusan memperkenalkan tiga hal paling mendasar dalam Python. Ketiganya dijelaskan pada *callout* di bawah — bacalah pelan-pelan di rumah, karena semuanya akan dipakai di **setiap** pertemuan setelah ini.

::: {.callout-note title="Konsep Python"}
### Variabel dan Penugasan Nilai {.unnumbered .unlisted}
`nama_praktikan = "Aulia"` artinya: siapkan sebuah kotak berlabel `nama_praktikan`, lalu masukkan tulisan `"Aulia"` ke dalamnya. Kotak berlabel itulah **variabel**, dan tanda `=` disebut **penugasan** (*assignment*) — perintah "isikan nilai di kanan ke dalam label di kiri", bukan pernyataan matematika "sama dengan".

Setelah diisi, label bisa dipakai di mana pun untuk memanggil isinya — itulah sebabnya `print(f"...{nama_praktikan}...")` bisa menampilkan `Aulia`. Isi kotak juga boleh diganti kapan saja dengan penugasan baru, misalnya `nama_praktikan = "Bima"`; isi lama tertimpa oleh yang baru.

Di studio, variabel seperti kotak berlabel di rak: sekali kotak "data survei" diisi, siapa pun yang butuh tinggal menyebut labelnya tanpa memindahkan isinya ke sana-sini. Nanti kita akan sering menaruh hal-hal panjang (misalnya *path* berkas data) ke dalam variabel pendek agar tidak lelah mengetik.
:::

::: {.callout-note title="Konsep Python"}
### Tipe Data Skalar — String dan Angka {.unnumbered .unlisted}
Perhatikan perbedaan dua macam isi kotak: `"Aulia"` ditulis **di antara tanda kutip**, sedangkan angka seperti `5` atau `2026` ditulis polos. Tanda kutip adalah penanda bahwa isinya adalah **teks** — Python menyebutnya **string** (`str`). Tanpa tanda kutip, Python membacanya sebagai **angka**: bilangan bulat (**integer**/`int`, misalnya `2026`) atau bilangan berkoma (**float**, misalnya `197.22` — perhatikan pemisah desimalnya titik, bukan koma).

Jenis-jenis isi ini disebut **tipe data**, dan yang berisi satu nilai tunggal seperti di atas disebut tipe data **skalar**. Tipe menentukan apa yang bisa dilakukan: angka bisa dihitung (`2 + 3` menghasilkan `5`), sedangkan string dirangkai (`"20" + "26"` menghasilkan `"2026"`, bukan `46`!). Salah membedakan keduanya adalah sumber kekeliruan yang sangat umum — kita akan bertemu pesan kesalahannya sebentar lagi.
:::

::: {.callout-note title="Konsep Python"}
### `print()` dan f-string {.unnumbered .unlisted}
`print()` adalah perintah untuk **menampilkan** sesuatu ke layar — petugas loket membacakan isi kotak yang kita sebut. Apa pun yang ditaruh di dalam tanda kurungnya akan dilaporkan ke terminal.

Bentuk `f"Halo, {nama_praktikan} dari {prodi}!"` disebut **f-string** (huruf `f` di depan tanda kutip artinya *formatted*). Di dalam f-string, bagian yang diapit kurung kurawal `{...}` akan **diganti dengan isi variabelnya** saat ditampilkan. Ini cara paling rapi untuk mencampur teks tetap dengan nilai variabel — tanpa f-string kita harus menyambung teks potong demi potong.
:::



### Membaca Pesan Kegagalan: Tiga *Error* Paling Umum dan *Debugging* Dasar

Cepat atau lambat — biasanya cepat — kode kita akan gagal berjalan dan Python menampilkan tulisan merah yang panjang. Jangan panik: pesan itu bukan kemarahan, melainkan **petugas loket yang balik bertanya** karena instruksi kita tidak bisa dia pahami atau kerjakan. Seperti peserta studio yang menerima catatan revisi dari asisten studio, perencana yang baik tidak takut pada catatan revisi; programmer yang baik tidak takut pada pesan *error*.

Kunci membacanya: lihat **baris paling bawah** terlebih dahulu. Di sanalah tertulis **nama jenis kesalahannya** beserta keterangan singkat, sedangkan baris-baris di atasnya (disebut *traceback*) menunjukkan di berkas dan baris ke berapa kesalahan terjadi. Tiga jenis inilah yang paling sering menyapa pemula — kita sengaja memproduksi ketiganya sekarang, supaya saat muncul sungguhan kita sudah kenal wajahnya.

**Wajah pertama: `SyntaxError` — kalimat instruksinya cacat.** Python punya tata bahasa. Lupa menutup tanda kutip atau tanda kurung membuat kalimat kita tidak selesai, dan Python menolak bahkan untuk mulai bekerja. Jalankan sel berikut yang lupa menutup tanda kutip dan kurung:

````python
print("Halo, Studio!
````

Perhatikan baris terakhirnya: nama kesalahannya `SyntaxError`, dan tanda `^` menunjuk lokasi Python kebingungan. Perbaikannya: lengkapi pasangan tanda yang hilang — setiap `"` dan `(` harus ada penutupnya.



**Wajah kedua: `NameError` — menyebut label kotak yang tidak ada.** Kesalahan ini muncul saat kita memanggil variabel yang belum pernah diisi, paling sering karena **salah ketik**. Di sel berikut, kita bermaksud memanggil `nama_praktikan` tetapi namanya kurang satu huruf:

````python
print(nama_praktika)
````

Python menjawab `NameError: name 'nama_praktika' is not defined` — "tidak ada kotak berlabel itu di rak". Perbaikannya: cocokkan ejaan nama variabel dengan yang didefinisikan, huruf demi huruf (huruf besar-kecil dibedakan!). Pada Python versi terbaru, sesekali pesan ini malah berbaik hati menebak maksud kita dengan saran `Did you mean: ...?`.



**Wajah ketiga: `TypeError` — menyuruh dua tipe data yang tak cocok bekerja sama.** Ingat pembahasan tipe data tadi: string dan angka punya kemampuan berbeda. Menyambung teks dengan angka memakai `+` membuat Python bingung harus menghitung atau merangkai:

````python
"Kota Bandar Lampung berusia " + 342
````

Baris terakhirnya `TypeError: can only concatenate str (not "int") to str` — "saya hanya bisa menyambung string dengan string, bukan dengan bilangan bulat". Perbaikannya yang paling rapi sudah kita kenal: bungkus dengan f-string, `print(f"Kota Bandar Lampung berusia {342}")`, dan Python otomatis mengubah angkanya menjadi teks.



Dari tiga perkenalan ini, lahirlah rutinitas ***debugging*** (mencari dan memperbaiki kesalahan) yang akan kita pakai sepanjang semester:



1. Baca **baris paling bawah**: apa nama jenis kesalahannya?



2. Lihat **nomor baris** yang ditunjuk *traceback*: di mana lokasinya?



3. Perbaiki satu hal itu saja, lalu **jalankan ulang**.



4. Masih gagal? Ulangi dari langkah 1. Setiap pesan baru adalah petunjuk baru.
